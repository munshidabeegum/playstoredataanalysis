```python
avg_rating=df.groupby("Category")["Rating"].mean()
avg_rating
```

```python
highest_Category=avg_rating.idxmax()
highest_value=avg_rating.max()
```

```python
print("Category with highest average rating:",highest_Category)
print("highest average rating:",highest_value)
```

```python
top_Categories=avg_rating.head(10)
plt.figure(figsize=(8,8))
plt.pie(top_Categories,
        labels=top_Categories.index,
        autopct="%1.1f%%",
        startangle=140)
plt.title("Top 10 app Categories by average rating")
plt.show()
```

```python
plt.figure()
sns.barplot(x="Type",y="Installs",data=df)
```

QUESTION AND ANSWERS

1.which categories of apps recieve the highest average rating?

```python
top_Categories=avg_rating.head(10)
plt.figure(figsize=(8,8))
plt.pie(top_Categories,
        labels=top_Categories.index,
        autopct="%1.1f%%",
        startangle=140)
plt.title("Top 10 app Categories by average rating")
plt.show()
```

Education,Art&Design,and Books and Reference often have the highest average rating

2.Do free apps consistently outperform paid apps in terms of installs?

```python
plt.figure()
sns.barplot(x="Type",y="Installs",data=df)
```

yes,free apps generally outperform paid apps in installs

3.Is there a strong connection between reviews and ratings?

```python
plt.figure()
sns.lineplot(x="Rating",y="Reviews",data=df)
```

There is no direct relationship between reviews and ratings.Apps with many reviews tend to be popular,but their ratings can still vary.

4.Which category dominates in terms of installs?

```python
Category_Installs=df.groupby("Category")["Installs"].sum().sort_values(ascending=False)
```

```python
plt.figure(figsize=(10,6))
sns.barplot(x=Category_Installs.values,y=Category_Installs.index)
plt.xlabel("Number of Installs")
plt.ylabel("Category")
plt.title("Total installs by category")
```

Games,communication and social dominate in installs
